```python
import numpy as np

my_student_number = 4402375

np.random.seed(my_student_number)

random_sample_size = 200
random_sample_creation = df.sample(n=random_sample_size, random_state=my_student_number)

print(f"Random Sample Created(200 observations):\n", random_sample_creation)
```

```python
import pandas as pd
from pygam import LinearGAM, s, f
import matplotlib.pyplot as plt

categorical_cols = ['rain', 'snow', 'Date', 'month', 'bridge']
df_encoded = pd.get_dummies(random_sample_creation, columns=categorical_cols, drop_first=True)

numeric_cols = ['High.Temp', 'Low.Temp', 'prec']
df_encoded[numeric_cols] = df_encoded[numeric_cols].apply(pd.to_numeric)

x= df_encoded.drop(columns=['counts'])
y= df_encoded['counts'] 

print(df_encoded)
```

```python
print("Column types for X:")
print(x.dtypes)

print("\nColumn type for y:")
print(y.dtypes)
```

```python
print("Checking for NaN values in x:")
print(x.isna().sum())  
print("Checking for Inf values in x:")
print((x == float('inf')).sum()) 

print("Checking for NaN values in y:")
print(y.isna().sum()) 
print("Checking for Inf values in y:")
print((y == float('inf')).sum()) 
```

```python
print(x['prec'])

print("Unique values in 'prec' column before cleaning:")
print(x['prec'].unique())
```

```python
print((x == np.inf).sum())

print(y.isna().sum())
```

```python
GAM_model = LinearGAM(
    s(0) +  
    s(1) +
    s(2) + 
    f(3) + 
    f(4)+
    f(5)+
    f(6) +
    f(7)  
).fit(x, y)

print(GAM_model.summary())
```
