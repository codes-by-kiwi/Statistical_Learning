Dropping the ID column

```python
# Drop the 'id' column
df.drop(columns=['id'], inplace=True)
print(df.head())
```

Splitting data into x and y data portions

Candidate models to compare performance when cv=1

```python
#candidate models to compare performance can be logistic regression, random forest and kNN
#since there are 31 predictor variables RF(random forest) would work

from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score


candidate_models = {
    "Logistic_Regression with c=0.5": LogisticRegression(C=0.5, random_state=42),
    "Logistic Regression with c=1.0": LogisticRegression(C=1.0, random_state=42),
    "Logistic Regression with c=10.0": LogisticRegression(C=10.0, random_state=42),
    "kNN (k=3)": KNeighborsClassifier(n_neighbors=3),
    "kNN (k=4)": KNeighborsClassifier(n_neighbors=4),
    "kNN (k=5)": KNeighborsClassifier(n_neighbors=5),
    "Random Forest (100 trees)": RandomForestClassifier(n_estimators=100, random_state=42),
    "Random Forest (500 trees)": RandomForestClassifier(n_estimators=500, random_state=42),
    "Random Forest (1000 trees)": RandomForestClassifier(n_estimators=1000, random_state=42)
}

candidate_performance = {}

for model_name, model in candidate_models.items():
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    accuracy = accuracy_score(y_test, y_pred)
    candidate_performance[model_name] = accuracy

for model_name, accuracy in candidate_performance.items():
    print(f"{model_name}: Accuracy = {accuracy:.4f}")
```

Candidate models to compare performance when cv=10

```python
from sklearn.model_selection import cross_val_score
import numpy as np

candidate_performance_cv_10 = {}  # Store individual fold scores for 10-fold CV

for model_name, model in candidate_models.items():
    cv_scores = cross_val_score(model, x_train, y_train, cv=10, scoring='accuracy')

    candidate_performance_cv_10[model_name] = cv_scores

    average_accuracy = np.mean(cv_scores)
    candidate_performance[model_name] = average_accuracy

for model_name, cv_scores in candidate_performance_cv_10.items():
    print(f"{model_name}: 10-fold CV Scores = {cv_scores}")

for model_name, accuracy in candidate_performance.items():
    print(f"{model_name}: Average Accuracy = {accuracy:.4f}")
```

Best performing model tested on test set and specificity, sensitivity and accuracy printed

```python
#The model we are going to select is Random forest with a 1000 trees because it performs the best on average on training set with cv validation
from sklearn.metrics import confusion_matrix, accuracy_score, recall_score

best_model = candidate_models["Random Forest (1000 trees)"]
best_model.fit(x_train, y_train)
y_pred_test = best_model.predict(x_test)
accuracy = accuracy_score(y_test, y_pred_test)
true_neg, false_pos, false_neg, true_pos = confusion_matrix(y_test, y_pred_test).ravel()

# Sensitivity (Recall) - Treat 'M' or malignant as the positive class
sensitivity = recall_score(y_test, y_pred_test, pos_label='M')

specificity = true_neg / (true_neg + false_pos)

print(f"Accuracy: {accuracy:.4f}")
print(f"Sensitivity (Recall): {sensitivity:.4f}")
print(f"Specificity: {specificity:.4f}")
```
