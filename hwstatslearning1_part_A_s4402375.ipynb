# Portioning dataset into train and test sets and Scaling of X predictors

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
X_train= train[['X1','X2','X3','X4','X5','X6']]
Y_train = train['Y'].values
X_test = test[['X1','X2','X3','X4','X5','X6']]
Y_test = test['Y'].values
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.model_selection import KFold


ten_fold_cv = KFold(n_splits=10, shuffle=True, random_state=42)
```

#Hyperparameter Grid search for Best K value in knn

```python
#hyperparameter grid search
k_value_set = range(1, 50)

mean_accuracies = []
```

```python

import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

for k in k_value_set:
    knn = KNeighborsClassifier(n_neighbors=k)
    cv_scores = cross_val_score(knn, X_train_scaled, Y_train, cv=ten_fold_cv, scoring='accuracy')
    mean_accuracies.append(np.mean(cv_scores))
```

```python
best_k = k_value_set[np.argmax(mean_accuracies)]
print(f"The Optimal value of K so far in my search is : {best_k}")
```

```python
knn_with_best_k = KNeighborsClassifier(n_neighbors=best_k)
knn_with_best_k.fit(X_train_scaled, Y_train)

test_accuracy = knn_with_best_k.score(X_test_scaled, Y_test)
print(f"Test Set Accuracy of the Optimal KNN Classifier: {test_accuracy:.4f}")

training_cv_accuracy = np.max(mean_accuracies)
print(f"Cross-validated accuracy of the optimal KNN classifier: {training_cv_accuracy:.4f}")
```

#Performing Lasso Logistic Regression with optimal lambda selection using cf validation

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LogisticRegressionCV
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
#over here I have kept 50 candidate values to keep the evaluation procedure similar to the KNN classifier
lasso_model = LogisticRegressionCV(Cs=50, penalty='l1', solver='liblinear', random_state=42) #values of C are evenly spaced on log scale
lasso_model.fit(X_train_scaled, Y_train)

best_lambda = 1/lasso_model.C_[0]
print(f"The Optimal value of lambda so far in my search is : {best_lambda}")
```

```python
Y_prediction = lasso_model.predict(X_test_scaled)
test_accuracy = accuracy_score(Y_test,Y_prediction)
print(f"Test Set Accuracy with Best Lambda: {test_accuracy:.4f}")
cv_accuracy = np.max(lasso_model.scores_[1].mean(axis=0))
print(f"Cross-validated Accuracy: {cv_accuracy:.4f}")
```

```python

lasso_model2 = LogisticRegressionCV(Cs=50, penalty='l2', solver='saga', random_state=42) #values of C are evenly spaced on log scale
lasso_model2.fit(X_train_scaled, Y_train)

best_lambda2 = 1/lasso_model2.C_[0]
print(f"The Optimal value of lambda so far in my search is : {best_lambda2}")
```

```python
Y_prediction2 = lasso_model2.predict(X_test_scaled)
test_accuracy2 = accuracy_score(Y_test,Y_prediction2)
print(f"Test Set Accuracy with Best Lambda: {test_accuracy2:.4f}")
cv_accuracy2 = np.max(lasso_model2.scores_[1].mean(axis=0))
print(f"Cross-validated Accuracy: {cv_accuracy2:.4f}")
```

#Selecting Optimal K for KNN classifier using CF validation - using all predictors




```python
from sklearn.model_selection import KFold
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

X_train_original= train.drop(columns=['Y'])
Y_train_original= train['Y'].values
X_test_original = test.drop(columns=['Y'])
Y_test_original = test['Y'].values

X_train_scaled_original = scaler.fit_transform(X_train_original)
X_test_scaled_original = scaler.transform(X_test_original)

ten_fold_cv_original = KFold(n_splits=10, shuffle=True, random_state=42)

#hyperparameter grid search
k_value_set_original = range(1, 50)
mean_accuracies_original = []

for k_original in k_value_set_original:
    knn_original = KNeighborsClassifier(n_neighbors=k_original)
    cv_scores_original = cross_val_score(knn_original, X_train_scaled_original, Y_train_original, cv=ten_fold_cv_original, scoring='accuracy')
    mean_accuracies_original.append(np.mean(cv_scores_original))

best_k_original = k_value_set_original[np.argmax(mean_accuracies_original)]
print(f"The Optimal value of K so far in my search in the original dataset with no X predictors removed is : {best_k_original}")

knn_with_best_k_original = KNeighborsClassifier(n_neighbors=best_k_original)
knn_with_best_k_original.fit(X_train_scaled_original, Y_train_original)

test_accuracy_original = knn_with_best_k_original.score(X_test_scaled_original, Y_test_original)
print(f"Test Set Accuracy of the Optimal KNN Classifier with no X predictors removed is: {test_accuracy_original:.4f}")

training_cv_accuracy_original = np.max(mean_accuracies_original)
print(f"Cross-validated accuracy of the optimal KNN classifier with no X predictors removed is: {training_cv_accuracy_original:.4f}")
```
